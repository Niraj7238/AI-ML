```python
import pandas as pd

```

```python
df = pd.read_csv("HousePricePrediction.csv")
```

```python
df.head()
df.shape
df.info()
df.describe()
```

```python
# DATA PREPROCESSING AND -- DATA CLEANING

# drop Id as it doesnot contribute in pricing
df.drop(['Id'],
axis = 1,
inplace = True)


# Replacing salePrice empty values with their mean value 
df['SalePrice'] = df['SalePrice'].fillna(
    df['SalePrice'].mean()
)

# Drop records with null values
df = df.dropna()
df.isnull().sum()
df.head()




```
